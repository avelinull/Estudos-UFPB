```python
#Exercicio 4 lista 4

while True:
    
    print("--- MENU DE OPÇÕES ---")
    print("1. Potência Elétrica Contínua (V x I)")
    print("2. Consumo de Energia (Potência x Tempo)")
    print("3. Sair")
    
    opcao = input("Digite a opção desejada: ").strip()

    if opcao == '1':
        print("[ Opção 1: Cálculo de Potência Elétrica ]")
        tensao = float(input("Digite a tensão (em Volts): "))
        corrente = float(input("Digite a corrente (em Amperes): "))
        potencia = tensao * corrente
        print(f"-> Potência Elétrica: {potencia:.2f} W")

    elif opcao == '2':
        print("[ Opção 2: Cálculo de Consumo de Energia ]")
        potencia = float(input("Digite a potência do aparelho (em Watts): "))
        tempo = float(input("Digite o tempo de uso diário (em horas): "))
        energia = (potencia * tempo) / 1000
        print(f"-> Energia consumida: {energia:.2f} kWh")

    elif opcao == '3':
        print("Programa encerrado com sucesso. Até logo!")
        break  

    else:
        print("Opção Inválida! Por favor, escolha uma opção entre 1 e 3.")
```
