```python
#Exercicio 3 lista 4

TOTAL_MODULOS = 25


temp_alta = 0


soma_potencia_faixa = 0.0
qtd_faixa_temp = 0


qtd_ef_pot_baixa = 0

for i in range(1, TOTAL_MODULOS + 1):
    print(f"--- Módulo Solar {i} de {TOTAL_MODULOS} ---")
    temperatura = float(input("Digite a temperatura de operação (°C): "))
    potencia = float(input("Digite a potência de pico gerada (W): "))
    eficiencia = float(input("Digite a eficiência medida (%): "))

    
    if temperatura > 45:
        temp_alta += 1

    
    if 25 <= temperatura <= 40:
        soma_potencia_faixa += potencia
        qtd_faixa_temp += 1

    
    if eficiencia < 17 and potencia < 300:
        qtd_ef_pot_baixa += 1


if qtd_faixa_temp > 0:
    media_potencia = soma_potencia_faixa / qtd_faixa_temp
else:
    media_potencia = 0.0


porcentagem_c = (qtd_ef_pot_baixa / TOTAL_MODULOS) * 100


print("================ RESUMO DA AVALIAÇÃO DA UFPB/CEAR ================")
print(f"a) Quantidade de módulos com temperatura > 45 °C: {temp_alta}")

if qtd_faixa_temp > 0:
    print(f"b) Média de potência (módulos entre 25°C e 40°C): {media_potencia:.2f} W")
else:
    print("b) Média de potência (módulos entre 25°C e 40°C): Nenhum módulo registrado nessa faixa.")

print(f"c) Porcentagem de módulos com eficiência < 17% e potência < 300 W: {porcentagem_c:.2f}%")
```
