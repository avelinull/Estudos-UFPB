```python
#Exercicio 1 lista 4

baixo = 0
moderado = 0
medio = 0
alto = 0
muito_alto = 0

TOTAL_RESIDENCIAS = 20


for i in range(1, TOTAL_RESIDENCIAS + 1):
    consumo = float(input(f"Digite o consumo da residência {i} (em kWh): "))
    
    
    if consumo <= 100:
        baixo += 1
    elif consumo <= 220:
        moderado += 1
    elif consumo <= 350:
        medio += 1
    elif consumo <= 500:
        alto += 1
    else:
        muito_alto += 1


porcentagem_baixo = (baixo / TOTAL_RESIDENCIAS) * 100
porcentagem_muito_alto = (muito_alto / TOTAL_RESIDENCIAS) * 100


print("--- RESULTADO DA ANÁLISE DE CONSUMO ---")
print("a) Quantidade de residências em cada faixa:")
print(f"   - 1ª Faixa (Baixo Consumo - Até 100 kWh): {baixo}")
print(f"   - 2ª Faixa (Consumo Moderado - 100 a 220 kWh): {moderado}")
print(f"   - 3ª Faixa (Consumo Médio - 220 a 350 kWh): {medio}")
print(f"   - 4ª Faixa (Consumo Alto - 350 a 500 kWh): {alto}")
print(f"   - 5ª Faixa (Muito Alto Consumo - Acima de 500 kWh): {muito_alto}")

print("b) Porcentagem em relação ao total (20 residências):")
print(f"   - Baixo Consumo: {porcentagem_baixo:.2f}%")
print(f"   - Muito Alto Consumo: {porcentagem_muito_alto:.2f}%")
```
