```python
#Exercicio 2 lista 4

momentos_turbinas = 5

turbina_desligada = 0
operaçao_normal = 0
operaçao_potencia = 0
vento_forte = 0

for i in range (1, momentos_turbinas+ 1):
    vento = float(input(f"Digite a velocidade do vento {i} (em m/s)"))

    if vento <= 3:
        turbina_desligada += 1
    elif vento > 3 and vento <= 12:
        operaçao_normal += 1
    elif vento > 12 and vento <= 25:
        operaçao_potencia += 1
    else:
        vento_forte += 1

baixa_porcentagem = (turbina_desligada/momentos_turbinas) * 100
alta_porcetagem = (vento_forte/momentos_turbinas) * 100

print(" --- RESULTADO DA ANÁLISE DA POTENCIA ---")
print("a) A quantidade de medições em cada faixa de operação da turbina: ")
print(f"   - 1ª Faixa (Por mais do vento ser fraco para operação, a força é): {turbina_desligada}")
print(f"   - 2ª Faixa (Operação padrão): {operaçao_normal}")
print(f"   - 3ª Faixa (Operação forte): {operaçao_potencia}")
print(f"   - 4ª Faixa (Pás desligadas por segurança): {vento_forte}")

print("b) Porcentagem das faixas com turbina desligada:")
print(f"   - Vento Muito Fraco: {baixa_porcentagem:.2f}%")
print(f"   - Vento Fortíssimo (Segurança): {alta_porcetagem:.2f}%")
```
