```python
#Exercicio 2 lista 4


TOTAL_ABASTECIMENTOS = 15


total_gasolina = 0.0
total_etanol = 0.0


for i in range(1, TOTAL_ABASTECIMENTOS + 1):
    print(f"--- Abastecimento {i} de {TOTAL_ABASTECIMENTOS} ---")
    tipo = input("Digite o tipo de combustível (G para Gasolina / E para Etanol): ").strip().upper()
    valor = float(input("Digite o valor abastecido (R$): "))

    
    if tipo == 'G':
        total_gasolina += valor
    elif tipo == 'E':
        total_etanol += valor
    else:
        print("Código de combustível inválido! O valor não foi contabilizado.")


total_geral = total_gasolina + total_etanol
parcela_gasolina = total_gasolina / 2


print("================ RESUMO DOS GASTOS ================")
print(f"a) Total gasto com Gasolina: R$ {total_gasolina:.2f}")
print(f"b) Total gasto com Etanol:   R$ {total_etanol:.2f}")
print(f"c) Total gasto geral:        R$ {total_geral:.2f}")
print(f"d) Valor de cada parcela da Gasolina (2x sem juros): R$ {parcela_gasolina:.2f}")
```
